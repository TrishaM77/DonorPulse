```python
df["interaction_date"] = pd.to_datetime(df["interaction_date"])
df["followup_due_date"] = pd.to_datetime(df["followup_due_date"])
df["donation_date"] = pd.to_datetime(df["donation_date"])
```

```python
df["converted"].mean()
```

```python
conversion_rate = df["converted"].mean() * 100
print("Conversion Rate:",round(conversion_rate, 2), "%")
```

```python
total_donations = df["donation_amount"].sum()
print("Total Donations: ", round(total_donations, 2))
```

```python
donors=df[df["converted"]==1]
average_donation = donors["donation_amount"].mean()
print("Average Donation Amount: ", round(average_donation, 2))
```

```python
contact_rate = (df["contact_status"] == 'Connected').mean() * 100
print("Contact Rate:", round(contact_rate, 2), "%")
```

```python
followups = df[df["followup_required"]==1]
```

```python
followup_rate=followups["followup_completed"].mean() * 100
print("Follow-up Completion Rate:", round(followup_rate, 2), "%")
```

```python
campaign_conversion = df.groupby("campaign_name")["converted"].mean() * 100
campaign_conversion
```

```python
campaign_conversion.sort_values(ascending=False)
```

```python
agent_conversion = df.groupby("agent_id")["converted"].mean() * 100

agent_conversion.sort_values(ascending=False)
```
